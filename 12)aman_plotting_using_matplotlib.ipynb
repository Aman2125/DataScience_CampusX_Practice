```python
# from a pandas dataframe
batsman = pd.read_csv('/content/sharma-kohli.csv')
batsman

plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit vs Virat run comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
```

```python
# colors(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'] , color = 'green') # color changed
plt.plot(batsman['index'],batsman['RG Sharma'] , color = 'black')

plt.title('Rohit vs Virat run comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
```

```python
from matplotlib.lines import lineStyles
plt.plot(batsman['index'],batsman['V Kohli'] , color = 'green' , linestyle='dashed') # different line style
plt.plot(batsman['index'],batsman['RG Sharma'] , color = 'black' , linestyle='dotted' , linewidth = 3) # line width change also

plt.title('Rohit vs Virat run comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
```

```python
from matplotlib.lines import lineStyles
plt.plot(batsman['index'],batsman['V Kohli'] , color = 'green' , linestyle='dashed' , marker='D') # diamond marker used
plt.plot(batsman['index'],batsman['RG Sharma'] , color = 'pink' , linestyle='dotted' , linewidth = 3 , marker='o')

plt.title('Rohit vs Virat run comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
```

```python
# legend -> location
from matplotlib.lines import lineStyles
plt.plot(batsman['index'],batsman['V Kohli'] , color = 'blue' , linestyle='dashed' , marker='D' , label ='Virat' ) # lebel used which can be called using function legend
plt.plot(batsman['index'],batsman['RG Sharma'] , color = 'red' , linestyle='dotted' , linewidth = 3 , marker='o' , label='Rohit')

plt.title('Rohit vs Virat run comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.legend()
```

```python
# limiting axes
price = [48000,54000,57000,49000,45000,43000,4500000] # outlier
year = [2016,2017,2018,2019,2020,2021,2022]
# because of outlier we cant see the graph variation , so using ylim to limit the axis
plt.plot(year , price)
plt.ylim(0,75000)
```

```python
# grid in graph
from matplotlib.lines import lineStyles
plt.plot(batsman['index'],batsman['V Kohli'] , color = 'blue' , linestyle='dashed' , marker='D' , label ='Virat' )
plt.plot(batsman['index'],batsman['RG Sharma'] , color = 'red' , linestyle='dotted' , linewidth = 3 , marker='o' , label='Rohit')

plt.title('Rohit vs Virat run comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.legend()
plt.grid() # to turn on the grid for better visualisation
```

```python
# show
from matplotlib.lines import lineStyles
plt.plot(batsman['index'],batsman['V Kohli'] , color = 'blue' , linestyle='dashed' , marker='D' , label ='Virat' )
plt.plot(batsman['index'],batsman['RG Sharma'] , color = 'red' , linestyle='dotted' , linewidth = 3 , marker='o' , label='Rohit')

plt.title('Rohit vs Virat run comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.legend()
plt.grid()

plt.show() # this function is used to show the graph , this must be used while plotting the graph in any other editor
```

### Bar chart

image.png

- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color='orange')
plt.xlabel('colors')
plt.ylabel('children')
```

```python
# horizontal bar chart
plt.barh(colors , children , color='orange')
```

```python
# color and label
df = pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0])-0.2, df['2015'] , width=0.2)
# to plot multiple bar side by side we will do some shortcuts by adjusting the width of the bars and using
#np.arange to produce numbers and then doing it
plt.bar(np.arange(df.shape[0]), df['2016'] , width=0.2)
plt.bar(np.arange(df.shape[0])+0.2, df['2017'] , width=0.2)
# now for showing the name of the batsman on x axis
plt.xticks(np.arange(df.shape[0]) , df['batsman'])
plt.show()
```

```python
# a problem
# simple bar chart
children = [10,20,40,10,30]
colors = ['red red red red red red ','blue blue blue blue blue blue','green','yellow','pink']
# as categories name is big so overlapping and to handle this we will use xticks
plt.bar(colors,children,color='orange')
plt.xticks(rotation = 'vertical')
plt.xlabel('colors')
plt.ylabel('children')
```

```python
# stacked bar chart
plt.bar(df['batsman'] , df['2015'] , label = '2015')
plt.bar(df['batsman'] , df['2016'] , bottom = df['2015'] , label='2016')
plt.bar(df['batsman'] , df['2017'] , bottom = df['2015']+df['2016'] , label = '2017')

plt.legend()
plt.show()
```

### Pie Chart

image.png

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
# simple data
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)

plt.show()
```

```python
df = pd.read_csv('/content/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'] , labels=df['batsman'] , autopct='%0.1f%%') # autopct is used to show the percentage
plt.show()
```

```python
# colors
plt.pie(df['batsman_runs'] , labels=df['batsman'] , autopct='%0.1f%%' , colors=['blue' , 'orange' , 'green' , 'cyan' , 'pink','brown'])
plt.show()
```

```python
# explode shadow
plt.pie(df['batsman_runs'] , labels=df['batsman'] , autopct='%0.1f%%' , explode=[0.1,0,0,0,0,0] , shadow=True)
plt.show()
```
