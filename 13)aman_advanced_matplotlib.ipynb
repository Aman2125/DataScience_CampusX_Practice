```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris = pd.read_csv('iris.csv')
iris.sample(5)
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'] , iris['PetalLengthCm'],c=iris['Species'] , cmap='jet' , alpha=0.7) # alpha tells about opacity of graph
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar() # gives side bar of color gradient
```

# Annotations

```python
batter = pd.read_csv('batter.csv')
batter.head()
```

```python
sample_df = batter.head(100).sample(25,random_state=5)
```

```python
sample_df
```

```python
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'] , sample_df['strike_rate'] , s=sample_df['runs'])
# annotating each point using loop
for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i] , sample_df['strike_rate'].values[i] , sample_df['batter'].values[i])
```

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.scatter(x,y)
# annotating points for more clarity using text function
plt.text(1,5,'Point 1')
plt.text(2,6,'Point 2')
plt.text(3,7,'Point 3')
plt.text(4,8,'Point 4' , fontdict={'size':12,'color':'brown'}) # some modification
```

# Horizontal and Vertical lines


```python
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'] , sample_df['strike_rate'] , s=sample_df['runs'])
# horizontal line for strike rate above 130
plt.axhline(130,color='red')

# vertical line for avg above 30
plt.axvline(30,color='red')
for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i] , sample_df['strike_rate'].values[i] , sample_df['batter'].values[i])
```

# Subplots
A different way to plot graphs.

```python
# till now we were doing
batter.sample()
```

```python
plt.scatter(batter['avg'] , batter['strike_rate'])
plt.title('Batter')
plt.xlabel('Avg')
plt.ylabel('strike_rate')

plt.show()
```

```python
# using different way
fig , ax = plt.subplots(figsize=(15,6)) # figsize=(15,6) using this inside for size increase or decrease
ax.scatter(batter['avg'] , batter['strike_rate'] , color='red' , marker = '+')
ax.set_title('Batter') # using set word for title and label
ax.set_xlabel('Avg')
ax.set_ylabel('strike_rate')

fig.show()

# the above method is used for subplottings.
```

```python
# eg plot runs vs avg and avg vs strike rate
fig , ax = plt.subplots(nrows=2 , ncols=1 , sharex=True) # 2 rows and 1 col , it can be anything according to use
# ax[0] , contains the first axis and ax[1] contains the other one
ax[0].scatter(batter['avg'] , batter['strike_rate'])
ax[1].scatter(batter['avg'] , batter['runs'])

ax[0].set_title('Avg vs Strike Rate')
ax[1].set_title('Avg vs Runs')

# sharex is used as x is same for both graphs
```

```python
# plotting 4 graphs
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].scatter(batter['avg'],batter['strike_rate'] , color='red')
ax[0,1].scatter(batter['avg'],batter['runs'])
ax[1,0].hist(batter['avg'])
ax[1,1].hist(batter['runs'])
```

```python
# another way but above one is better
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1) # means 2 row, 2 col, 1st graph
ax1.scatter(batter['avg'],batter['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batter['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batter['avg'])
```

# 3d scatter plots


```python
fig = plt.figure()

ax = plt.subplot(projection='3d')
ax.scatter3D(batter['avg'] , batter['strike_rate'] , batter['runs'])
ax.set_title('IPL batsman analysis')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike Rate')
ax.set_zlabel('Runs')
```

# 3D surface plots

```python
# used in loss functions
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx , yy = np.meshgrid(x,y)
z= xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx, yy ,z , cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy) # some other function
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx, yy ,z , cmap='viridis')
fig.colorbar(p)
```

# Contour Plots


```python
# used to convert 3d graphs into 2d ones
fig = plt.figure()
ax = plt.subplot()

p = ax.contour(xx, yy ,z , cmap='viridis')
fig.colorbar(p)
```

```python
# used to convert 3d graphs into 2d ones
fig = plt.figure()
ax = plt.subplot()

p = ax.contourf(xx, yy ,z , cmap='viridis')
fig.colorbar(p)
```

# Heatmap

```python
# find how much sixes have been hit on each ball in each over of the ipl till now
delivery = pd.read_csv('/content/IPL_Ball_by_Ball_2008_2022.csv')
delivery.sample()
```

```python
temp_df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run']==6)]
temp_df.head()
```

```python
grid = temp_df.pivot_table(index = 'overs' , columns='ballnumber' , values='batsman_run' , aggfunc='count')
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid) # imshow used for heatmap
plt.yticks(delivery['overs'].unique() , list(range(1,21)))
plt.xticks(np.arange(0,6), list(range(1,7)))
plt.colorbar()
```
